## How it works
The simulator makes a representation of **four home sensors (motion, temperature, door and smoke)**, first as dictionary in the code and then as objects of a class named `Sensor´. The simulator generates random readings for each minute.

The reading is passed to `process_reading`, using the **if/else logic** to check the **three type of alerts (security, safety and comfort)**. The security alerts is MOTION or OPEN DOOR while the house is in AWAY or SLEEP mode. The safety refers to extreme temperatures or smoke, in any home mode. The comfort refers to temperature outside in HOME mode. The result is displayed in severity level, high or critical alerts as notifications to the owner.

- **Safety**: temperature below 35 °F or above 95 °F, or smoke detected (any mode) | CRITICAL

```python
#STEP 1: Import necessary libraries
import random
from datetime import datetime
 
# System configuration
system_mode = ["HOME", "AWAY", "SLEEP"]
alert_severities = ["LOW", "MEDIUM", "HIGH", "CRITICAL"]
#SENSOR_TYPES = ["MOTION", "DOOR", "WINDOW", "SMOKE", "HOT", "COLD"] - use that later on

 
# Current system state
current_mode = "AWAY"


#STEP 2: Define Data Structures
##Create a function that initializes sensor data structures
##Define the structure for storing sensor information
##Define the structure for storing alert information

def create_sensor (sensor_id, location, sensor_type, threshold=None):
    sensor = {
        "id": sensor_id,
        "location": location,
        "type": sensor_type,
        "threshold": threshold,
        "current_value": None
    }
    return sensor

def create_alert(severity, message, sensor_id, timestamp):
    alert = {
        "severity": severity,
        "message": message,
        "sensor_id": sensor_id,
        "timestamp": timestamp
    }
    return alert

# Initialize sensors for the Peterson home
sensors = [
    create_sensor("MOTION_001", "Living Room", "motion"), #motion for the living room 
    create_sensor("TEMP_001", "Kitchen", "temperature", threshold=35), #temperature for the kitchen with threshold
    create_sensor("DOOR_001", "Front Door", "door"), #door for the front door
    create_sensor("SMOKE_001", "Bedroom", "smoke")  #smoke for the bedroom 
]

# Initialize alerts for the Peterson home
alerts = [
    create_alert("HIGH", "Motion detected in the Living Room", "MOTION_001", datetime.now()),
    create_alert("CRITICAL", "Temperature exceeded threshold in the Kitchen", "TEMP_001", datetime.now()),
    create_alert("MEDIUM", "Front Door opened unexpectedly", "DOOR_001", datetime.now()),
    create_alert("CRITICAL", "Smoke detected in the Bedroom", "SMOKE_001", datetime.now())
]

#Verify the sensors created:

print(f"Initialized {len(sensors)} sensors")
for sensor in sensors:
    print(f"  - {sensor['id']}: {sensor['location']} ({sensor['type']})")


#STEP 3: If-Else Logic
##Create a function that checks if a sensor reading is abnormal
##Use if/else statements to compare values against thresholds
##Check the system mode to determine if alerts should be triggered

def is_abnormal_reading(sensor, reading_value):
    sensor_type = sensor["type"]
    #temperature sensor logic
    if sensor_type == "temperature": 
        if reading_value < 35:  #example threshold for temperature
            return True
        elif reading_value > 95:  #example threshold for temperature
            return True
        else:
            return False
    #motion sensor logic
    elif sensor_type == "motion":
        if reading_value: True  #example condition for motion
        else:
            return False
    #door sensor logic
    elif sensor_type == "door":
        if reading_value == "OPEN":  #example condition for door
            return True
        else:
            return False 
    #smoke sensor logic
    elif sensor_type == "smoke":
        if reading_value == "DETECTED":  #example condition for smoke
            return True
        else:
            return False
    return True

def should_trigger_security_alert(sensor, reading_value, system_mode):
    if system_mode == "AWAY":
        if sensor["type"] in ["motion", "door"]:
            return is_abnormal_reading(sensor, reading_value)
        else:
            return False
    elif system_mode == "HOME":
             return False
    elif system_mode == "SLEEP":
            if sensor["type"] in ["motion", "door"]:
                return is_abnormal_reading(sensor, reading_value)
            else:
                return False
    return False

#Test of functions (temperature and security)

# Test temperature check
test_sensor = create_sensor("TEMP_TEST", "Test Room", "temperature", threshold=35)
print(f"34°F is abnormal: {is_abnormal_reading(test_sensor, 34)}")  # Should be True
print(f"68°F is abnormal: {is_abnormal_reading(test_sensor, 68)}")  # Should be False
 
# Test security alert
motion_sensor = create_sensor("MOTION_TEST", "Test Room", "motion")
print(f"Motion in AWAY mode triggers alert: {should_trigger_security_alert(motion_sensor, True, 'AWAY')}")  # Should be True
print(f"Motion in HOME mode triggers alert: {should_trigger_security_alert(motion_sensor, True, 'HOME')}")  # Should be False


#STEP 4: Building Functions
##Write a function to generate realistic sensor readings
##Create a function to process readings and determine if alerts are needed
##Implement a function to trigger and display alerts
##Add a logging function to record all events


def generate_reading(sensor): #the reading for the sensor
    if sensor["type"] == "temperature":
        return random.randint(30,100)
    elif sensor["type"] == "motion":
        return random.choice([True, False])
    elif sensor["type"] == "door":
        return random.choice(["OPEN","CLOSED"])
    elif sensor["type"] == "smoke":
        return random.choice(["CLEAR","DETECTED"])


for sensor in sensors:
    print(sensor["type"], "-", generate_reading(sensor))


def process_reading(sensor, reading_value, system_mode):
    alerts = []                                       
    timestamp = datetime.now().strftime("%H:%M:%S")
    if should_trigger_security_alert(sensor, reading_value, system_mode):
        alerts.append(
            create_alert(
                "HIGH",
                f"{sensor['location']} triggered while in {system_mode} mode!",
                sensor["id"],
                timestamp
            )
        )
    if sensor["type"] == "temperature" and reading_value < 35:
        alerts.append(create_alert(
            "CRITICAL",
            f"SAFETY: {sensor['location']} temperature {reading_value} ºF - FROZEN pipe risk!",
            sensor["id"],
            timestamp
        ))
    if sensor["type"] == "temperature" and reading_value > 95:
            alerts.append(create_alert(
                "CRITICAL",
                f"SAFETY: {sensor['location']} temperature {reading_value} ºF - HOT pipe risk!",
                sensor["id"],
                timestamp
            ))
    if sensor["type"] == "smoke" and reading_value == "DETECTED":
        alerts.append(create_alert(
             "CRITICAL",
             f"SAFETY: {sensor['location']} {reading_value} - FIRE risk!",
             sensor["id"],
             timestamp
            ))
    if system_mode == "HOME" and sensor["type"] == "temperature":
        if reading_value < 65 or reading_value > 75:
            alerts.append(create_alert(
                "LOW",
                f"COMFORT: {sensor['location']} temperature {reading_value}°F is outside the comfort range",
                sensor["id"],
                timestamp 

            ))
    return alerts

def trigger_alert(alert):
    severity_symbol = {
        "LOW": "ℹ️ ",
        "MEDIUM": "⚠️ ",
        "HIGH": "🚨 ",
        "CRITICAL": "🔥 "
    }
    symbol = severity_symbol.get(alert["severity"], "⚠️ ")
    print(f"[ALERT!] {symbol} {alert['severity']}: {alert['message']}")


def log_event(message, timestamp=None):
    if timestamp is None:
        timestamp = datetime.now().strftime("%H:%M:%S")
    print(f"[LOG] [{timestamp}] {message}")


# Test reading generation
test_sensor = sensors[0]  # Motion sensor
reading = generate_reading(test_sensor)
print(f"Generated reading for {test_sensor['location']}: {reading}")
 
# Test processing
alerts = process_reading(test_sensor, True, "AWAY")
if alerts:
    trigger_alert(alerts[0])


#STEP 5: Creating Classes - be more organized
##Define a Sensor class with an __init__ method
##Add methods: read(), isAbnormal(), and reset()
##Update your code to use the Sensor class instead of dictionaries

class Sensor:
    def __init__(self, sensor_id, location, sensor_type, threshold=None):
        self.id = sensor_id
        self.location = location
        self.type = sensor_type
        self.threshold = threshold
        self.current_value = None

    def read(self):
        reading = generate_reading({"type": self.type})
        self.current_value = reading
        return reading

    def isAbnormal(self):
        return is_abnormal_reading({"type": self.type}, self.current_value)

    def reset(self):
        self.current_value = None

    def __str__(self):
        status = "No reading" if self.current_value is None else str(self.current_value)
        return f"{self.id} ({self.location}): {status}"

sensor_objects = [
    Sensor("MOTION_001", "Living Room", "motion"),
    Sensor("TEMP_001", "Kitchen", "temperature", threshold=35),
    Sensor("DOOR_001", "Front Door", "door"),
    Sensor("SMOKE_001", "Bedroom", "smoke")
]


test_sensor = Sensor("TEST_001", "Test Room", "temperature", threshold=35)
test_sensor.read()
print(f"Sensor reading: {test_sensor.current_value}")
print(f"Is abnormal: {test_sensor.isAbnormal()}")
print(f"Sensor info: {test_sensor}")

def run_simulation(duration_minutes=5, system_mode="AWAY"):
    print("=" * 50)
    print("=== HomeGuard Security System ===")
    print("=" * 50)
    print(f"Mode: {system_mode}\n")

    sensors = [
        Sensor("MOTION_001", "Living Room", "motion"),
        Sensor("TEMP_001", "Kitchen", "temperature", threshold=35),
        Sensor("DOOR_001", "Front Door", "door"),
        Sensor("SMOKE_001", "Bedroom", "smoke")
    ]

    for minute in range(duration_minutes):
        current_time = datetime.now().strftime("%H:%M:%S")
        print(f"\nTime: {current_time}")

        for sensor in sensors:
            reading = sensor.read()

            if sensor.type == "temperature":
                status = "Normal" if 65 <= reading <= 75 else "Abnormal"
                print(f"[READING] {sensor.location} Temperature: {reading}°F ({status})")
            elif sensor.type == "motion":
                status = "DETECTED" if reading else "No activity"
                print(f"[READING] {sensor.location} Motion: {status}")
            elif sensor.type == "door":
                print(f"[READING] {sensor.location}: {reading}")
            elif sensor.type == "smoke":
                print(f"[READING] {sensor.location} Smoke: {reading}")

            sensor_dict = {
                "id": sensor.id,
                "location": sensor.location,
                "type": sensor.type,
                "threshold": sensor.threshold
            }
            alerts = process_reading(sensor_dict, reading, system_mode)

            for alert in alerts:
                trigger_alert(alert)
                if alert["severity"] in ["HIGH", "CRITICAL"]:
                    log_event("Sending notification to homeowner...")

        import time
        time.sleep(0.5)

    print("\n" + "=" * 50)
    print("Simulation complete!")
    print("=" * 50)


if __name__ == "__main__":
    run_simulation(duration_minutes=3, system_mode="AWAY")
```
